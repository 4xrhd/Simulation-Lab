## 1. Advanced Scalar & Array Operations


```python
import numpy as np

scalar = 4.1
arr_1d = np.array([-0.9, -0.4, 0.0, 0.25, 0.6, 0.85, 1.5, 3.2], dtype=np.float64)

print("Scalar value:", scalar)
print("Original 1D Array:", arr_1d)
```

```python
# Arithmetic Operations
print("--- Arithmetic Operations ---")
print("Addition with scalar (arr + scalar):\n", arr_1d + scalar)
print("\nSubtraction with scalar (arr - scalar):\n", arr_1d - scalar)
print("\nMultiplication with scalar (arr * scalar):\n", arr_1d * scalar)
print("\nDivision with scalar (arr / scalar):\n", arr_1d / scalar)

#pwer with bases
power_result = np.power(np.abs(arr_1d), scalar)
print("\nPower operation (|arr|^scalar):\n", power_result)
```

```python
print("--- Mathematical Functions ---")

abs_val = np.abs(arr_1d)
print("Absolute Value (np.abs):\n", abs_val)

sqrt_val = np.sqrt(np.abs(arr_1d))
print("\nSquare Root (np.sqrt of absolute values):\n", sqrt_val)

# exponential function
exp_val = np.exp(arr_1d)
print("\nExponential Function (np.exp):\n", exp_val)

# natural flogarithm 
log_val = np.log(np.abs(arr_1d) + 1e-10)
print("\nNatural Logarithm (np.log):\n", log_val)

#log base 10
log10_val = np.log10(np.abs(arr_1d) + 1e-10)
print("\nLogarithm Base 10 (np.log10):\n", log10_val)
```

```python
# trigonometric Functions
print("--- Trigonometric Functions ---")
print("sin():\n", np.sin(arr_1d))
print("\ncos():\n", np.cos(arr_1d))
print("\ntan():\n", np.tan(arr_1d))

arr_clipped = np.clip(arr_1d, -1.0, 1.0)
print("\nClipped Array for Inverse Trig (-1 to 1):\n", arr_clipped)

print("\narcsin():\n", np.arcsin(arr_clipped))
print("\narccos():\n", np.arccos(arr_clipped))
print("\narctan():\n", np.arctan(arr_1d))
```

## 2. Matrix Mathematical Operations


```python

matrix_4x3 = np.array([
    [-1.25,  0.50,  2.10],
    [-0.75, -2.80,  0.00],
    [ 0.35,  1.20, -0.95],
    [ 3.50, -1.10,  0.80]
], dtype=np.float64)

print("Custom 4x3 Matrix:\n", matrix_4x3)
print("Matrix Shape:", matrix_4x3.shape)
```

```python
# Exponential & Logarithmic Operations
print("--- Exponential & Logarithmic Operations ---")
print("exp():\n", np.exp(matrix_4x3))

# Handling negative values properly using np.abs() and epsilon offset
print("\nlog() (Natural Logarithm of absolute values):\n", np.log(np.abs(matrix_4x3) + 1e-10))
print("\nlog10() (Logarithm Base 10 of absolute values):\n", np.log10(np.abs(matrix_4x3) + 1e-10))
```

```python
# Other Mathematical Operations
print("--- Other Mathematical Operations ---")
print("Absolute Value:\n", np.abs(matrix_4x3))
print("\nSquare Root (of absolute values):\n", np.sqrt(np.abs(matrix_4x3)))
print("\nSquare:\n", np.square(matrix_4x3))
print("\nCube Root (handles negatives natively):\n", np.cbrt(matrix_4x3))

print("\n--- Rounding and Remainder Operations ---")
print("Remainder when divided by 3 (matrix % 3):\n", np.fmod(matrix_4x3, 3))
print("\nround():\n", np.round(matrix_4x3))
print("\nfloor():\n", np.floor(matrix_4x3))
print("\nceil():\n", np.ceil(matrix_4x3))
```

## 3. Matrix Statistics and Analysis


```python
# Create a 3x4 matrix with random floating-point values
np.random.seed(42)  # Set seed for reproducible results
matrix_3x4 = np.random.uniform(-10, 10, (3, 4))

print("Random 3x4 Matrix:\n", matrix_3x4)
```

```python
# Basic Statistics
print("--- Basic Statistics ---")
max_val = np.max(matrix_3x4)
min_val = np.min(matrix_3x4)
argmax_flat = np.argmax(matrix_3x4)
argmin_flat = np.argmin(matrix_3x4)
argmax_2d = np.unravel_index(argmax_flat, matrix_3x4.shape)
argmin_2d = np.unravel_index(argmin_flat, matrix_3x4.shape)

print(f"Maximum Value: {max_val:.4f}")
print(f"Minimum Value: {min_val:.4f}")
print(f"Index Position of Max Value (Flat Index: {argmax_flat}, 2D Index: {argmax_2d})")
print(f"Index Position of Min Value (Flat Index: {argmin_flat}, 2D Index: {argmin_2d})")
print(f"Sum of All Elements: {np.sum(matrix_3x4):.4f}")
print(f"Product of All Elements: {np.prod(matrix_3x4):.4e}")
print(f"Mean: {np.mean(matrix_3x4):.4f}")
print(f"Median: {np.median(matrix_3x4):.4f}")
print(f"Variance: {np.var(matrix_3x4):.4f}")
print(f"Standard Deviation: {np.std(matrix_3x4):.4f}")
```

```python
# Row and Column Analysis
print("--- Row and Column Analysis ---")
row_sum = np.sum(matrix_3x4, axis=1)
col_sum = np.sum(matrix_3x4, axis=0)
row_max = np.max(matrix_3x4, axis=1)
col_min = np.min(matrix_3x4, axis=0)

print("Sum of each row (axis=1):\n", row_sum)
print("\nSum of each column (axis=0):\n", col_sum)
print("\nMaximum value from each row (axis=1):\n", row_max)
print("\nMinimum value from each column (axis=0):\n", col_min)
```

## 4. Matrix Sorting and Searching Operations


```python
# Create a 4x4 integer matrix using random values
np.random.seed(101)
matrix_4x4 = np.random.randint(-20, 20, (4, 4))

print("Original 4x4 Integer Matrix:\n", matrix_4x4)
```

```python
# Sorting Operations
print("--- Sorting Operations ---")
# 1. Sort each row in ascending order
row_asc = np.sort(matrix_4x4, axis=1)
print("1. Sort each row in ascending order:\n", row_asc)

# 2. Sort each row in descending order
row_desc = np.sort(matrix_4x4, axis=1)[:, ::-1]
print("\n2. Sort each row in descending order:\n", row_desc)

# 3. Sort entire matrix in ascending order
flat_asc = np.sort(matrix_4x4.flatten())
matrix_asc = flat_asc.reshape(4, 4)
print("\n3. Sort entire matrix in ascending order (reshaped 4x4):\n", matrix_asc)

# 4. Sort entire matrix in descending order
flat_desc = flat_asc[::-1]
matrix_desc = flat_desc.reshape(4, 4)
print("\n4. Sort entire matrix in descending order (reshaped 4x4):\n", matrix_desc)
```

```python
print("--- Finding Elements ---")
matrix_flat = matrix_4x4.flatten()

# 5 largest and 5 smallest values
five_largest = np.sort(matrix_flat)[-5:][::-1]
five_smallest = np.sort(matrix_flat)[:5]

print("Five largest values in matrix:", five_largest)
print("Five smallest values in matrix:", five_smallest)

# mean
mat_mean = matrix_4x4.mean()
print(f"Matrix Mean: {mat_mean:.4f}")

#position
pos_gt_mean = np.argwhere(matrix_4x4 > mat_mean)
print("Positions (Row, Col) of elements greater than mean:\n", pos_gt_mean)

# counts of positive, negative, and zero values
pos_count = np.sum(matrix_4x4 > 0)
neg_count = np.sum(matrix_4x4 < 0)
zero_count = np.sum(matrix_4x4 == 0)

print(f"Count of positive values (>0): {pos_count}")
print(f"Count of negative values (<0): {neg_count}")
print(f"Count of zero values (==0): {zero_count}")
```
