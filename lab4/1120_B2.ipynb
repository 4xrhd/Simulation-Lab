```python
NormalValues = np.random.normal(9, 2, 1000)

print(np.mean(NormalValues))
print(np.mean(NormalValues))

print(np.std(NormalValues))

print(np.max(NormalValues))

# Create a histogram
plt.hist(NormalValues, bins=30, edgecolor='black')
plt.title('Histogram of Normal Distribution (mean=9, std=2)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```
