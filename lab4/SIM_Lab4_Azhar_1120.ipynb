## 3. Uniform Distribution with Random Numbers
A **uniform distribution** is a probability distribution where every outcome in a given range $[a, b]$ is equally likely.
- **Probability Density Function (PDF):** $f(x) = \frac{1}{b - a}$ for $a \le x \le b$
- **Mean:** $\mu = \frac{a + b}{2} = 0.5$ for $U[0, 1)$
- **Variance:** $\sigma^2 = \frac{(b - a)^2}{12} = \frac{1}{12} \approx 0.0833$

### 3.1 Uniform Distribution with Raw Frequency ($N = 100,000$)

```python
# Generate 100,000 random numbers in [0, 1)
random_numbers = np.random.rand(100000)

# Plot raw frequency histogram
plt.figure(figsize=(8, 4.5))
plt.hist(random_numbers, bins=50, alpha=0.75, color='royalblue', edgecolor='white')
plt.xlabel('Value')
plt.ylabel('Count (Raw Frequency)')
plt.title('Histogram of Uniformly Distributed Random Numbers (Raw Frequency)', fontsize=12, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### 3.2 Uniform Distribution with Normalized Probability Density (`density=True`)

```python
# Plot normalized density histogram
plt.figure(figsize=(8, 4.5))
plt.hist(random_numbers, density=True, bins=50, alpha=0.6, color='dodgerblue', edgecolor='black')
plt.axhline(1.0, color='crimson', linestyle='--', linewidth=2, label='Theoretical PDF f(x)=1.0')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.title('Histogram of Uniformly Distributed Random Numbers (Normalized Density)', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 4. Normal (Gaussian) Distribution
A **normal distribution** is a symmetric, continuous bell-shaped distribution where values cluster symmetrically around the mean $\mu$.
- **Probability Density Function (PDF):**
  $$f(x) = \frac{1}{\sigma \sqrt{2\pi}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$
- **Standard Normal Distribution:** $Z \sim \mathcal{N}(0, 1)$ where $\mu = 0$ and $\sigma = 1$.

### 4.1 Standard Normal Distribution $\mathcal{N}(0, 1)$ ($N = 10,000$)

```python
# Generate 10,000 random values from standard normal distribution
SNormalValues = np.random.randn(10000)  # mean=0, std=1

print(f'Empirical Sample Mean: {np.mean(SNormalValues):.4f} (Theoretical: 0.0)')
print(f'Empirical Sample Std:  {np.std(SNormalValues):.4f} (Theoretical: 1.0)')

# Create a histogram
plt.figure(figsize=(8, 4.5))
plt.hist(SNormalValues, bins=30, density=True, color='mediumseagreen', edgecolor='black', alpha=0.7)
plt.title('Standard Normal Distribution $\mathcal{N}(0, 1)$', fontsize=12, fontweight='bold')
plt.xlabel('Value (Z-score)')
plt.ylabel('Probability Density')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### 4.2 Parameterized Normal Distribution $\mathcal{N}(\mu=100, \sigma=10)$ ($N = 10,000$)

```python
# Generate 10,000 random values from normal distribution with mean=100 and std=10
NormalValues = np.random.normal(100, 10, 10000)

print(f'Empirical Sample Mean: {np.mean(NormalValues):.4f} (Theoretical: 100.0)')
print(f'Empirical Sample Std:  {np.std(NormalValues):.4f} (Theoretical: 10.0)')

# Create a histogram
plt.figure(figsize=(8, 4.5))
plt.hist(NormalValues, bins=30, density=True, color='coral', edgecolor='black', alpha=0.7)
plt.title('Parameterized Normal Distribution $\mathcal{N}(\mu=100, \sigma=10)$', fontsize=12, fontweight='bold')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Student Work 2: Random Integers with Seed Control
- Create a $3 \times 3$ matrix of random integers between $1$ and $50$.
- Run the code twice:
  - **With a fixed seed (`np.random.seed(10)`)** to ensure reproducibility.
  - **Without setting a seed**, allowing random values to change each time.
- Observe and explain the differences in outputs.

```python
# Execution 1 with Seed 10
np.random.seed(10)
mat_seeded_run1 = np.random.randint(1, 51, (3, 3))
print('Run 1 with np.random.seed(10):\n', mat_seeded_run1)

# Execution 2 with Seed 10
np.random.seed(10)
mat_seeded_run2 = np.random.randint(1, 51, (3, 3))
print('\nRun 2 with np.random.seed(10):\n', mat_seeded_run2)

# Check equivalence of seeded runs
print('\nAre Seeded Run 1 and Run 2 identical?:', np.array_equal(mat_seeded_run1, mat_seeded_run2))

# Execution 1 without fixed seed
np.random.seed(None)
mat_unseeded_run1 = np.random.randint(1, 51, (3, 3))
print('\nRun 1 without seed:\n', mat_unseeded_run1)

# Execution 2 without fixed seed
np.random.seed(None)
mat_unseeded_run2 = np.random.randint(1, 51, (3, 3))
print('\nRun 2 without seed:\n', mat_unseeded_run2)

# Check equivalence of unseeded runs
print('\nAre Unseeded Run 1 and Run 2 identical?:', np.array_equal(mat_unseeded_run1, mat_unseeded_run2))
```

### Observations and Explanation
1. **With Seed (`np.random.seed(10)`):**
   - When the PRNG seed is fixed to `10`, the generator is initialized with the exact same deterministic state vector.
   - Consequently, `Run 1` and `Run 2` produce **identical $3 \times 3$ matrices** with identical values in every cell.
   - This is crucial for **debugging**, **unit testing**, and **reproducible simulation experiments**.

2. **Without Seed (`np.random.seed(None)`):**
   - When no seed is specified, NumPy derives entropy from the operating system (e.g., current time clock or `/dev/urandom`).
   - As a result, each successive execution yields a **different, stochastic random matrix**.

3. **Small Sample Variance vs. Large Sample Asymptotics:**
   - In Task 1.1, the 10-element uniform sample does not appear flat because a small sample size has high empirical sampling variance.
   - In Task 1.2, with $N = 500$, the standard normal sample clearly resembles the theoretical Gaussian bell curve centered near $0$ with standard deviation near $1.0$.
