### Step 1: Importing Required Python Packages

### Task 1: Scalar & Array Operations
Here I am using a scalar value of `3.7` and a 6-element floating point array `[1.5, 3.2, 5.0, 7.8, 9.1, 10.4]` to test various element-wise mathematical functions.

```python
# Define scalar parameter and custom 6-element array
scalar_val = 3.7
data_arr = np.array([1.5, 3.2, 5.0, 7.8, 9.1, 10.4])

# Trigonometric functions
sin_out = np.sin(data_arr)
cos_out = np.cos(data_arr)
tan_out = np.tan(data_arr)

# Safe inverse trig functions using domain clipping to [-1, 1]
normalized_arr = data_arr / np.max(data_arr)
asin_out = np.arcsin(np.clip(normalized_arr, -1.0, 1.0))
acos_out = np.arccos(np.clip(normalized_arr, -1.0, 1.0))
atan_out = np.arctan(data_arr)

# Exponential and natural logarithm
exp_out = np.exp(data_arr)
log_out = np.log(data_arr)

# Absolute and square root values
abs_out = np.abs(data_arr)
sqrt_out = np.sqrt(data_arr)

# Modulo / remainder using scalar divisor
rem_out = np.remainder(data_arr, scalar_val)

# Integer rounding functions
round_out = np.round(data_arr)
floor_out = np.floor(data_arr)
ceil_out = np.ceil(data_arr)

# Displaying formatted results
print("--- Task 1 Results ---")
print("Scalar:", scalar_val)
print("Input Array:", data_arr)
print("\nSin:", sin_out)
print("Cos:", cos_out)
print("Tan:", tan_out)
print("Arcsin (clipped):", asin_out)
print("Arccos (clipped):", acos_out)
print("Arctan:", atan_out)
print("Exponential:", exp_out)
print("Natural Log:", log_out)
print("Absolute:", abs_out)
print("Square Root:", sqrt_out)
print(f"Remainder (divided by {scalar_val}):", rem_out)
print("Rounded:", round_out)
print("Floor:", floor_out)
print("Ceil:", ceil_out)
```

### Task 2: Mathematical Operations on a 3x2 Matrix
Creating a 3x2 matrix with custom values to evaluate trigonometric, logarithmic, exponential, and rounding functions across two dimensions.

```python
# Define custom 3x2 matrix
mat_3x2 = np.array([
    [2.4, 5.7],
    [8.1, 3.6],
    [6.3, 9.5]
])

# Compute element-wise trig functions
sin_m = np.sin(mat_3x2)
cos_m = np.cos(mat_3x2)
tan_m = np.tan(mat_3x2)

# Inverse trig functions with domain clipping [-1.0, 1.0]
norm_mat = mat_3x2 / np.max(mat_3x2)
asin_m = np.arcsin(np.clip(norm_mat, -1.0, 1.0))
acos_m = np.arccos(np.clip(norm_mat, -1.0, 1.0))
atan_m = np.arctan(mat_3x2)

# Exp, Log, Abs, Sqrt
exp_m = np.exp(mat_3x2)
log_m = np.log(mat_3x2)
abs_m = np.abs(mat_3x2)
sqrt_m = np.sqrt(mat_3x2)

# Remainder when divided by 2
rem_m = np.remainder(mat_3x2, 2)

# Rounding functions
round_m = np.round(mat_3x2)
floor_m = np.floor(mat_3x2)
ceil_m = np.ceil(mat_3x2)

print("--- Task 2: 3x2 Matrix Analysis ---")
print("Original Matrix:\n", mat_3x2)
print("\nSin:\n", np.round(sin_m, 4))
print("\nCos:\n", np.round(cos_m, 4))
print("\nTan:\n", np.round(tan_m, 4))
print("\nArcsin:\n", np.round(asin_m, 4))
print("\nArccos:\n", np.round(acos_m, 4))
print("\nArctan:\n", np.round(atan_m, 4))
print("\nExponential:\n", np.round(exp_m, 4))
print("\nLogarithm:\n", np.round(log_m, 4))
print("\nAbsolute:\n", abs_m)
print("\nSquare Root:\n", np.round(sqrt_m, 4))
print("\nRemainder (% 2):\n", np.round(rem_m, 4))
print("\nRound:\n", round_m)
print("\nFloor:\n", floor_m)
print("\nCeil:\n", ceil_m)
```

### Task 3: Vector Functions & Statistics on a 2x3 Matrix
Applying summary statistics, sorting mechanisms (row-wise and entire matrix descending), and min/max indexing on a 2x3 matrix with floating point values.

```python
# Create a 2x3 matrix with float numbers
mat_2x3 = np.array([
    [4.2, 8.7, 1.5],
    [9.3, 3.1, 6.4]
])

# Finding Max and Min values alongside their flat and 2D indices
max_val = np.max(mat_2x3)
max_flat_idx = np.argmax(mat_2x3)
max_2d_idx = np.unravel_index(max_flat_idx, mat_2x3.shape)

min_val = np.min(mat_2x3)
min_flat_idx = np.argmin(mat_2x3)
min_2d_idx = np.unravel_index(min_flat_idx, mat_2x3.shape)

# Total length / element count of matrix
total_len = mat_2x3.size

# Row-wise descending sort (sort axis=1 then reverse columns)
row_desc = np.sort(mat_2x3, axis=1)[:, ::-1]

# Entire matrix descending sort (flatten, sort, reverse)
entire_desc = np.sort(mat_2x3.flatten())[::-1]

# Basic summary metrics
total_sum = np.sum(mat_2x3)
total_product = np.prod(mat_2x3)
med_val = np.median(mat_2x3)
avg_val = np.mean(mat_2x3)
std_dev = np.std(mat_2x3)

print("--- Task 3: 2x3 Matrix Operations ---")
print("Input Matrix:\n", mat_2x3)
print(f"\nMaximum Value: {max_val} at 2D Position (row, col): {max_2d_idx}")
print(f"Minimum Value: {min_val} at 2D Position (row, col): {min_2d_idx}")
print(f"Matrix Element Count (Length): {total_len}")
print("\nRow-wise Descending Sort:\n", row_desc)
print("Entire Matrix Descending Sort:", entire_desc)
print(f"\nSum: {total_sum:.2f}")
print(f"Product: {total_product:.2f}")
print(f"Median: {med_val:.2f}")
print(f"Mean: {avg_val:.2f}")
print(f"Standard Deviation: {std_dev:.4f}")
```

### Bonus Task: Random Matrix Generation & Seaborn Visualization
Using `np.random.uniform()` to dynamically generate matrices and plotting heatmaps with `seaborn` and `matplotlib`.

```python
# Set random seed so results stay reproducible
np.random.seed(101)

# Generate random float matrices
random_mat_3x2 = np.round(np.random.uniform(2.0, 15.0, size=(3, 2)), 2)
random_mat_2x3 = np.round(np.random.uniform(2.0, 15.0, size=(2, 3)), 2)

print("Generated Random 3x2 Matrix:\n", random_mat_3x2)
print("\nGenerated Random 2x3 Matrix:\n", random_mat_2x3)

# Set up figure layout for heatmaps
plt.figure(figsize=(11, 4))

# Heatmap 1: 3x2 Matrix
plt.subplot(1, 2, 1)
sns.heatmap(random_mat_3x2, annot=True, cmap="Blues", fmt=".2f", cbar=True)
plt.title("Heatmap of Random 3x2 Matrix")
plt.xlabel("Columns")
plt.ylabel("Rows")

# Heatmap 2: 2x3 Matrix
plt.subplot(1, 2, 2)
sns.heatmap(random_mat_2x3, annot=True, cmap="Greens", fmt=".2f", cbar=True)
plt.title("Heatmap of Random 2x3 Matrix")
plt.xlabel("Columns")
plt.ylabel("Rows")

plt.tight_layout()
plt.show()
```
