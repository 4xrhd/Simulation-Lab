## 1. Necessary Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# seed for reproducibility
np.random.seed(42)
```

## 2. Matrix Types & Fundamental Analysis

```python
# Special matrix types
identity_mat = np.eye(4)
zero_mat = np.zeros((3, 3))
ones_mat = np.ones((3, 4))
diag_mat = np.diag([1, 2, 3])
upper_tri = np.triu([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
lower_tri = np.tril([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
random_mat = np.random.rand(3, 3)

print("Identity Matrix (4x4):\n", identity_mat)
print("\nZero Matrix (3x3):\n", zero_mat)
print("\nOnes Matrix (3x4):\n", ones_mat)
print("\nDiagonal Matrix:\n", diag_mat)
print("\nUpper Triangular Matrix:\n", upper_tri)
print("\nLower Triangular Matrix:\n", lower_tri)
print("\nRandom Matrix (3x3):\n", random_mat)
```

```python
# Matrix properties and linear algebra functions
A = np.array([[1, 2], [3, 4]], dtype=float)

size_A = A.shape
det_A = np.linalg.det(A)
rank_A = np.linalg.matrix_rank(A)
eigenvalues, eigenvectors = np.linalg.eig(A)
char_poly = np.poly(A)
roots = np.roots(char_poly)

if det_A != 0:
    inv_A = np.linalg.inv(A)
else:
    inv_A = "Matrix is singular, cannot compute inverse"

print("Size of A:", size_A)
print("Determinant of A:", det_A)
print("Inverse of A:\n", inv_A)
print("Rank of A:", rank_A)
print("Eigenvalues of A:", eigenvalues)
print("Eigenvectors of A:\n", eigenvectors)
print("Characteristic polynomial coefficients:", char_poly)
print("Roots of Characteristic Polynomial:", roots)
```

## Task 1: Generate and Plot Two Random Vectors

```python
# Create two vectors with 15 random floats
vector1 = np.random.rand(15)
vector2 = np.random.rand(15)
indices = np.arange(1, 16)

# Plot vectors on the same graph
plt.figure(figsize=(10, 5))
plt.plot(indices, vector1, marker='o', linestyle='-', color='b', label='Vector 1')
plt.plot(indices, vector2, marker='s', linestyle='--', color='r', label='Vector 2')

plt.title('Comparison of Two Random Vectors (15 Floats Each)')
plt.xlabel('Vector Element Index (1 to 15)')
plt.ylabel('Float Value')
plt.xticks(indices)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```

## Task 2: Matrix Operations and Visualization (4x4 Matrices)

```python
# Create two 4x4 matrices with random integer values
A = np.random.randint(1, 10, size=(4, 4))
B = np.random.randint(1, 10, size=(4, 4))

print("Matrix A (4x4):\n", A)
print("\nMatrix B (4x4):\n", B)

# Perform matrix operations
matrix_add = A + B
matrix_sub = A - B
matrix_mult = np.dot(A, B)  # compatible 4x4 multiplication

print("\nMatrix Addition (A + B):\n", matrix_add)
print("\nMatrix Subtraction (A - B):\n", matrix_sub)
print("\nMatrix Multiplication (A @ B):\n", matrix_mult)
```

```python
# Visualize results of each operation using bar plots
# Each bar represents one element of the 4x4 matrix (16 elements total)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
indices = range(1, 17)

# Bar plot for Addition
axes[0].bar(indices, matrix_add.flatten(), color='royalblue', edgecolor='black')
axes[0].set_title('Matrix Addition (4x4 Elements)')
axes[0].set_xlabel('Element Index (1 to 16)')
axes[0].set_ylabel('Value')
axes[0].set_xticks(indices)
axes[0].grid(True, linestyle=':', alpha=0.6)

# Bar plot for Subtraction
axes[1].bar(indices, matrix_sub.flatten(), color='seagreen', edgecolor='black')
axes[1].set_title('Matrix Subtraction (4x4 Elements)')
axes[1].set_xlabel('Element Index (1 to 16)')
axes[1].set_ylabel('Value')
axes[1].set_xticks(indices)
axes[1].grid(True, linestyle=':', alpha=0.6)

# Bar plot for Multiplication
axes[2].bar(indices, matrix_mult.flatten(), color='crimson', edgecolor='black')
axes[2].set_title('Matrix Multiplication (4x4 Elements)')
axes[2].set_xlabel('Element Index (1 to 16)')
axes[2].set_ylabel('Value')
axes[2].set_xticks(indices)
axes[2].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 3. Additional Matrix Visualizations (Heatmap & Line Graph)
