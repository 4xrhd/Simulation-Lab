## 1. Imports

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# set seed for reproducible results
np.random.seed(42)
```

## 2. Special Matrix Construction & Properties

```python
# Special matrix types
identity_mat = np.eye(4)
zero_mat = np.zeros((3, 3))
ones_mat = np.ones((3, 4))
diag_mat = np.diag([1, 2, 3])
upper_tri = np.triu([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
lower_tri = np.tril([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
random_mat = np.random.rand(3, 3)

print("Identity Matrix (4x4):\n", identity_mat)
print("\nZero Matrix (3x3):\n", zero_mat)
print("\nOnes Matrix (3x4):\n", ones_mat)
print("\nDiagonal Matrix:\n", diag_mat)
print("\nUpper Triangular Matrix:\n", upper_tri)
print("\nLower Triangular Matrix:\n", lower_tri)
print("\nRandom Matrix (3x3):\n", random_mat)
```

## 3. Matrix Analysis (Determinant, Rank, Eigenvalues, Inverse & Characteristic Polynomial)

```python
# Sample matrix A
A = np.array([[1, 2], [3, 4]], dtype=float)

shape_A = A.shape
det_A = np.linalg.det(A)
rank_A = np.linalg.matrix_rank(A)
eigenvalues, eigenvectors = np.linalg.eig(A)
char_poly = np.poly(A)
roots = np.roots(char_poly)

if det_A != 0:
    inv_A = np.linalg.inv(A)
else:
    inv_A = "Matrix is singular, cannot compute inverse"

print("Size of A:", shape_A)
print("Determinant of A:", det_A)
print("Rank of A:", rank_A)
print("Inverse of A:\n", inv_A)
print("\nEigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
print("\nCharacteristic Polynomial Coeffs:", char_poly)
print("Roots of Characteristic Polynomial:", roots)
```

## 4. Matrix Arithmetic Operations

```python
# Create two random 3x3 matrices
A_ops = np.random.randint(0, 10, size=(3, 3))
B_ops = np.random.randint(0, 10, size=(3, 3))

matrix_sum = A_ops + B_ops
matrix_diff = A_ops - B_ops
matrix_prod = np.dot(A_ops, B_ops)

print("Matrix A:\n", A_ops)
print("\nMatrix B:\n", B_ops)
print("\nSum (A + B):\n", matrix_sum)
print("\nDifference (A - B):\n", matrix_diff)
print("\nProduct (A @ B):\n", matrix_prod)
```

## 5. Line Graph of Flattened Matrices

```python
A_flat = A_ops.flatten()
B_flat = B_ops.flatten()
x = np.arange(len(A_flat))

plt.figure(figsize=(8, 4))
plt.plot(x, A_flat, marker='o', linestyle='-', label="Matrix A")
plt.plot(x, B_flat, marker='s', linestyle='--', label="Matrix B")
plt.xlabel("Flat Index")
plt.ylabel("Value")
plt.title("Line Graph of Flattened Matrices A and B")
plt.legend()
plt.grid(True)
plt.show()
```

## 6. Vector Comparison Plots (15 Random Floats)

```python
vector1 = np.random.rand(15)
vector2 = np.random.rand(15)
indices = np.arange(1, 16)

# Line comparison plot
plt.figure(figsize=(10, 4))
plt.plot(indices, vector1, marker='o', linestyle='-', color='b', label='Vector 1')
plt.plot(indices, vector2, marker='s', linestyle='--', color='r', label='Vector 2')
plt.title('Comparison of Two Random Vectors (15 Floats Each)')
plt.xlabel('Element Index')
plt.ylabel('Value')
plt.xticks(indices)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()

# Scatter plot
plt.figure(figsize=(6, 4))
plt.scatter(vector1, vector2, color='purple', edgecolors='black', s=80)
plt.title('Scatter Plot: Vector 1 vs Vector 2')
plt.xlabel('Vector 1 Values')
plt.ylabel('Vector 2 Values')
plt.grid(True, linestyle=':', alpha=0.7)
plt.tight_layout()
plt.show()
```

## 7. Heatmap of 3x3 Matrix

```python
heatmap_mat = np.random.rand(3, 3)

plt.figure(figsize=(5, 4))
sns.heatmap(heatmap_mat, annot=True, cmap='coolwarm', linewidths=0.5, linecolor='black', fmt='.2f')
plt.title('Heatmap of 3x3 Random Matrix')
plt.show()
```

## 8. Subplots of Matrix Operations (Bar Charts)

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].bar(range(1, 10), matrix_sum.flatten(), color='royalblue')
axes[0].set_title('Matrix Addition')
axes[0].set_xlabel('Flat Index')
axes[0].set_ylabel('Value')

axes[1].bar(range(1, 10), matrix_diff.flatten(), color='seagreen')
axes[1].set_title('Matrix Subtraction')
axes[1].set_xlabel('Flat Index')
axes[1].set_ylabel('Value')

axes[2].bar(range(1, 10), matrix_prod.flatten(), color='crimson')
axes[2].set_title('Matrix Multiplication')
axes[2].set_xlabel('Flat Index')
axes[2].set_ylabel('Value')

plt.tight_layout()
plt.show()
```
