```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set seed for reproducible outputs
np.random.seed(42)
```

# Task 1:
Generate and Plot Two Random Vectors:
Create two vectors, each containing 15 random floats of your choice.
Plot these vectors on the same graph using matplotlib to compare how they look.
Label the axes appropriately and give the plot a title.


```python
# Create two 15-element vectors of random floats
v1 = np.random.rand(15)
v2 = np.random.rand(15)
idx = np.arange(1, 16)

# Plot vectors on the same graph
plt.figure(figsize=(9, 4.5))
plt.plot(idx, v1, marker='o', linestyle='-', color='blue', label='Vector 1')
plt.plot(idx, v2, marker='s', linestyle='--', color='red', label='Vector 2')

plt.title('Comparison of Two 15-Element Random Vectors')
plt.xlabel('Vector Element Index (1 to 15)')
plt.ylabel('Float Value')
plt.xticks(idx)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```

# task2
Matrix Operations and Visualization:
Create two 4x4 matrices with random values of your choice.
Perform the following operations:
Matrix Addition
Matrix Subtraction
Matrix Multiplication (ensure matrices are compatible for multiplication)
For each of these operations, visualize the result using bar plots. Each bar should represent one element of the matrix.


```python
# Create two 4x4 matrices with random integer values
A = np.random.randint(1, 10, size=(4, 4))
B = np.random.randint(1, 10, size=(4, 4))

print("Matrix A (4x4):\n", A)
print("\nMatrix B (4x4):\n", B)

# Perform matrix operations
add_res = A + B
sub_res = A - B
mult_res = np.dot(A, B)  # 4x4 matrix multiplication

print("\nMatrix Addition (A + B):\n", add_res)
print("\nMatrix Subtraction (A - B):\n", sub_res)
print("\nMatrix Multiplication (A * B):\n", mult_res)
```

```python
# Visualize results using bar plots (1 bar per element of the 4x4 matrix)
fig, axes = plt.subplots(1, 3, figsize=(18, 4.5))
bars_idx = range(1, 17)

# Addition Bar Plot
axes[0].bar(bars_idx, add_res.flatten(), color='royalblue', edgecolor='black')
axes[0].set_title('Matrix Addition (4x4 Elements)')
axes[0].set_xlabel('Element Index (1 to 16)')
axes[0].set_ylabel('Value')
axes[0].set_xticks(bars_idx)
axes[0].grid(True, linestyle=':', alpha=0.6)

# Subtraction Bar Plot
axes[1].bar(bars_idx, sub_res.flatten(), color='seagreen', edgecolor='black')
axes[1].set_title('Matrix Subtraction (4x4 Elements)')
axes[1].set_xlabel('Element Index (1 to 16)')
axes[1].set_ylabel('Value')
axes[1].set_xticks(bars_idx)
axes[1].grid(True, linestyle=':', alpha=0.6)

# Multiplication Bar Plot
axes[2].bar(bars_idx, mult_res.flatten(), color='crimson', edgecolor='black')
axes[2].set_title('Matrix Multiplication (4x4 Elements)')
axes[2].set_xlabel('Element Index (1 to 16)')
axes[2].set_ylabel('Value')
axes[2].set_xticks(bars_idx)
axes[2].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
